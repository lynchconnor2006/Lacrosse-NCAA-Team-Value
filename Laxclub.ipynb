```python
import json, re
import pandas as pd

with open("Cumulative Efficiency – D1 Men – Lacrosse Reference.html", encoding="utf-8") as f:   # change the path
    html = f.read()

m = re.search(r"var\s+td\s*=\s*", html)
td, _ = json.JSONDecoder().raw_decode(html[m.end():])

rows = [{
    "team": t["display_name"],
    "off_efficiency": t["off_efficiency"]["val"],
    "def_efficiency": t["def_efficiency"]["val"],
    "cumulative_efficiency": t["adj_off_efficiency"]["val"],
} for t in td["data"]]

df = pd.DataFrame(rows).sort_values("off_efficiency", ascending=False).reset_index(drop=True)
df.index += 1

# convert decimals to percentages (0.386 -> 38.6)
for col in ["off_efficiency", "def_efficiency", "cumulative_efficiency"]:
    df[col] = (df[col] * 100).round(1)
    
print(df.head(15))
df.to_csv("d1_men_efficiency.csv", index_label="off_eff_rank")
```
